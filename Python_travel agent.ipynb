```python
print("=== Travel Agent ===")

destination = input("Destination (Pakistan/Turkey/Maldives): ").lower()
activity = input("Activity (mountains/beach/city): ").lower()

if destination == "pakistan" and activity == "mountains":
    print("Recommendation: Hunza")
elif destination == "pakistan" and activity == "beach":
    print("Recommendation: Karachi")
elif destination == "turkey" and activity == "city":
    print("Recommendation: Istanbul")
elif destination == "maldives" and activity == "beach":
    print("Recommendation: Maldives")
else:
    print("No suitable destination found.")
```
